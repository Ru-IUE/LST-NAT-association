Original entry name kept with an execution-order prefix. Uses the corrected shared methods under `00-Support/lstnat`; stored outputs are cleared.


```python
from pathlib import Path
import subprocess
import sys
ROOT = next((p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
             if (p / "00-Support" / "lstnat").is_dir()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside Upload Code")
sys.path.insert(0, str(ROOT / "00-Support"))
from lstnat.settings import load, output_path
CONFIG = ROOT / "00-Support" / "config" / "local.json"
if not CONFIG.exists():
    CONFIG = ROOT / "00-Support" / "config" / "example.json"
CFG = load(CONFIG)
def run_command(*args):
    return subprocess.run([sys.executable, str(ROOT / "00-Support" / "run.py"), *args,
                           "--config", str(CONFIG)], cwd=ROOT, check=True)
def run_script(relative):
    return subprocess.run([sys.executable, str(ROOT / relative), "--config", str(CONFIG)],
                          cwd=ROOT, check=True)
```

Train the T1 model using the architecture and optimization settings from Supplement S3. Requires TensorFlow and complete 10 m embedding imagery. The same 20% subset monitors early stopping; it is not an independent test set.


```python
run_command("train", "--time", "T1")
```
